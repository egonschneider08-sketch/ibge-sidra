```python

import sys
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

# Localiza a raiz do projeto
projeto = Path.cwd()

if projeto.name.lower() == "notebooks":
    projeto = projeto.parent

sys.path.insert(0, str(projeto))

from SRC.api import consultar_sidra
from SRC.tratamento import tratar_dados
```

```python

dados = consultar_sidra()

resultados = tratar_dados(dados)

df = pd.DataFrame(resultados)

df
```

```python

plt.figure(figsize=(11, 5))

plt.plot(
    df["periodo"],
    df["indice"],
    marker="o",
    linewidth=2
)

plt.axhline(
    y=100,
    linestyle="--",
    label="Base 2022 = 100"
)

plt.title("Produção Industrial Brasileira - IBGE")
plt.xlabel("Período")
plt.ylabel("Número-índice (2022 = 100)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python

plt.figure(figsize=(11, 5))

plt.bar(
    df["periodo"],
    df["diferenca_base"]
)

plt.axhline(y=0, linewidth=1)

plt.title("Diferença em relação à base de 2022")
plt.xlabel("Período")
plt.ylabel("Diferença (%)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python

print("RESUMO DA PRODUÇÃO INDUSTRIAL")
print("-" * 35)

print(f"Maior índice: {df['indice'].max():.2f}")
print(f"Menor índice: {df['indice'].min():.2f}")
print(f"Média dos índices: {df['indice'].mean():.2f}")

print(
    f"Último índice consultado: "
    f"{df['indice'].iloc[-1]:.2f}"
)
```
